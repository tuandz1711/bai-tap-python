```python
# Bai 4.1
import math

class HinhCau:
    def __init__(self, ban_kinh):
        if ban_kinh <= 0:
            raise ValueError("Bán kính phải lớn hơn 0")
        self.ban_kinh = ban_kinh

    def dien_tich(self):
        return 4 * math.pi * self.ban_kinh ** 2

    def the_tich(self):
        return (4 / 3) * math.pi * self.ban_kinh ** 3

    def __str__(self):
        return f"Hình cầu có bán kính {self.ban_kinh}"


# r = 2
try:
    hinh_cau = HinhCau(2)

    print(hinh_cau)
    print("Diện tích:", hinh_cau.dien_tich())
    print("Thể tích:", hinh_cau.the_tich())

except ValueError as e:
    print(e)


# r = -1
try:
    hinh_cau = HinhCau(-1)
    print(hinh_cau)
    print("Diện tích:", hinh_cau.dien_tich())
    print("Thể tích:", hinh_cau.the_tich())

except ValueError as e:
    print(e)
```

```python
# Bai 4.3
class SoDuKhongDu(Exception):
    pass

class TaiKhoan:
    def __init__(self, chu_tai_khoan, so_du=0):
        if so_du < 0:
            raise ValueError("Số dư ban đầu không được âm")

        self.chu_tai_khoan = chu_tai_khoan
        self.so_du = so_du
        self.lich_su = []

    def nap(self, tien):
        if tien <= 0:
            raise ValueError("Số tiền nạp phải lớn hơn 0")

        self.so_du += tien
        self.lich_su.append(f"Nạp {tien} đồng")
        return self.so_du

    def rut(self, tien):
        if tien <= 0:
            raise ValueError("Số tiền rút phải lớn hơn 0")

        if tien > self.so_du:
            raise SoDuKhongDu("Số dư không đủ để rút tiền")

        self.so_du -= tien
        self.lich_su.append(f"Rút {tien} đồng")
        return self.so_du

    def __str__(self):
        return f"Chủ tài khoản: {self.chu_tai_khoan}, Số dư: {self.so_du} đồng"
# Kiem tra
tk = TaiKhoan("Nguyen Van A", 1000000)
# Nap 500000
tk.nap(500000)
print("Số dư: ",tk.so_du )
# Rut 200000
tk.rut(200000)
print("Số dư: ",tk.so_du )
#Rut 2000000
try:
    tk.rut(2000000)
except SoDuKhongDu as e:
    print("Lỗi:", e)

print("Lịch sử giao dịch:", tk.lich_su)

print("Kiểm tra bất biến:", tk.so_du >= 0)
```

```python
# Bai 4.5
import math

class HinhHoc:
    def dien_tich(self):
        raise NotImplementedError("Lop con phai cai dat dien_tich()")

    def __str__(self):
        return "Hinh hoc"


class HinhTron(HinhHoc):
    def __init__(self, r):
        self.r = r

    def dien_tich(self):
        return math.pi * self.r ** 2

    def __str__(self):
        return f"Hinh tron ban kinh {self.r}"


class HinhChuNhat(HinhHoc):
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def dien_tich(self):
        return self.a * self.b

    def __str__(self):
        return f"Hinh chu nhat {self.a} x {self.b}"


class HinhTamGiac(HinhHoc):
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def dien_tich(self):
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))

    def __str__(self):
        return f"Hinh tam giac ({self.a}, {self.b}, {self.c})"


# Tao danh sach hinh
ds = [
    HinhTron(1),
    HinhChuNhat(3, 4),
    HinhTamGiac(3, 4, 5)
]

# Sap xep theo dien tich tang dan
ds.sort(key=lambda h: h.dien_tich())

# In danh sach
tong = 0

for h in ds:
    print(h, "- Dien tich:", round(h.dien_tich(), 3))
    tong += h.dien_tich()

print("Tong dien tich:", tong)
print("Gia tri pi + 18:", math.pi + 18)
```

```python
# Bai 4.7
import pandas as pd
df = pd.read_csv("goldprice.csv")

# Chuyen cot Date sang dang thoi gian
df["Date"] = pd.to_datetime(df["Date"], format="%Y-%m")

# Lay nam
df["Year"] = df["Date"].dt.year

# Tinh gia trung binh tung nam
gia_tb = df.groupby("Year").agg(
    Gia_TB=("Price", "mean"),
    So_Thang=("Price", "count")
)

# Chi lay cac nam du 12 thang
gia_tb = gia_tb[gia_tb["So_Thang"] == 12]

# Tinh phan tram tang giam so voi nam truoc
gia_tb["Tang_Giam"] = gia_tb["Gia_TB"].pct_change() * 100

# Loai bo nam khong co nam truoc de so sanh
ket_qua = gia_tb.dropna()

# Sap xep theo muc tang giam
tang_manh = ket_qua.sort_values(
    by="Tang_Giam", ascending=False
)

giam_manh = ket_qua.sort_values(
    by="Tang_Giam", ascending=True
)

# In 3 nam tang manh nhat
print("Ba nam tang manh nhat:")
print(tang_manh.head(3)[["Gia_TB", "Tang_Giam"]])

# In nam giam manh nhat
print("\nNam giam manh nhat:")
print(giam_manh.head(1)[["Gia_TB", "Tang_Giam"]])
```

```python
# Bai 4.9
import csv

class DongKhongHopLe(Exception):
    pass


# Tao tep luong
with open("luong.csv", "w", newline="", encoding="utf-8") as f:
    f.write("nam_kn,luong\n")
    f.write("1,30\n")
    f.write("2,35\n")
    f.write("abc,40\n")
    f.write("3,\n")
    f.write("4,50\n")
    f.write("5,55\n")
# Doc tep va xu ly du lieu
ds_hop_le = []
ds_loi = []

with open("luong.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)

    for so_dong, dong in enumerate(reader, start=2):
        try:
            nam_kn = int(dong["nam_kn"])
            luong = float(dong["luong"])

            ds_hop_le.append({
                "nam_kn": nam_kn,
                "luong": luong
            })

        except (ValueError, TypeError):
            try:
                raise DongKhongHopLe("Dong khong hop le")
            except DongKhongHopLe:
                ds_loi.append(so_dong)

# In ket qua
print("Cac dong hop le:")
for dong in ds_hop_le:
    print(dong)
print("Danh sach dong loi:", ds_loi)
# Tinh luong trung binh
if len(ds_hop_le) > 0:
    tong = 0
    for dong in ds_hop_le:
        tong += dong["luong"]
    print("Luong trung binh:", tong / len(ds_hop_le))
```

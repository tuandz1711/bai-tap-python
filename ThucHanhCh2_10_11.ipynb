```python
#Bai 2.3
import math
luong_gio = 100000
gio_chuan = 40

while True:
    s = input("Nhap so gio lam trong tuan: ")
    try:
        gio = float(s)
        if gio < 0:
            print("Loi: so gio khong duoc am")
        else:
            break
    except ValueError:
        print("Loi: ban phai nhap mot so")

if gio <= gio_chuan:
    luong = gio * luong_gio
else:
    gio_vuot = gio - gio_chuan
    luong = gio_chuan * luong_gio + gio_vuot * luong_gio * 1.5

print("Luong tuan cua ban la:", luong, "dong")
```

```python
#Bai 2.4
so_luong = 0
tong = 0
lon_nhat = 0
nho_nhat = 0

while True:
    s = input("nhap so thuc: ")
    if s == "xong"2:
        break

    try:
        x = float(s)
    except ValueError:
        print("gia tri khong phai so ", s)
        continue

    if so_luong == 0:
        lon_nhat = x
        nho_nhat = x
    else:
        if x > lon_nhat:
            lon_nhat = x
        if x < nho_nhat:
            nho_nhat = x

    so_luong = so_luong + 1
    tong = tong + x

if so_luong == 0:
    print("chua co du lieu")
else:
    trung_binh = tong / so_luong
    print("so luong hop le:", so_luong)
    print("tong:", tong)
    print("trung binh:", trung_binh)
    print("gia tri lon nha:", lon_nhat)
    print("gia tri nho nhat:", nho_nhat)
```

```python
#Bai 2.7
n = 27
so_buoc = 0
lon_nhat = n

while n != 1:
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1

    so_buoc = so_buoc + 1

    if n > lon_nhat:
        lon_nhat = n

print("sp buoc la:", so_buoc)
print("so lon nhat:", lon_nhat)

```

```python
#Bai 2.10
import random

random.seed(2026)
danh_sach_N = [100, 10000, 1000000]

for N in danh_sach_N:
    so_diem_trong = 0

    for i in range(N):
        x = random.random()
        y = random.random()
        if x * x + y * y <= 1:
            so_diem_trong = so_diem_trong + 1

    pi_uoc_luong = 4 * so_diem_trong / N
    sai_so = abs(pi_uoc_luong - math.pi)

    print("N =", N)
    print("  pi uoc luong:", pi_uoc_luong)
    print("  sai so tuyet doi:", sai_so)


    assert sai_so < 6 / math.sqrt(N)
```

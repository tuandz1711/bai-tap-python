```python
#Bai 3.4
import statistics

ds = [(1, 4.0), (2, 5.5), (3, 6.0), (4, 7.5), (5, 8.0)]

x = [a for a, b in ds]
y = [b for a, b in ds]

x_tb = sum(x) / len(x)
y_tb = sum(y) / len(y)

sxy = 0
sxx = 0

for i in range(len(x)):
    sxy += (x[i] - x_tb) * (y[i] - y_tb)
    sxx += (x[i] - x_tb) ** 2

b = sxy / sxx
a = y_tb - b * x_tb

print("a =", a)
print("b =", b)
print("Phương trình: y =", a, "+", b, "x")
print("Dự đoán khi x = 6:", a + b * 6)

print("Statistics:", statistics.linear_regression(x, y))
```

```python
#BAi 3.2
diem = [7.5, 9.0, 3.0, 8.0, 6.5, 10.0, 7.0, 2.5]
n = 2

diem_sap_xep = sorted(diem)

diem_con_lai = diem_sap_xep[n:len(diem_sap_xep) - n]

tb_con_lai = sum(diem_con_lai) / len(diem_con_lai)

tb_diem = sum(diem) / len(diem)

print("Danh sách gốc:", diem)
print("Các điểm còn lại:", diem_con_lai)
print("Trung bình thông thường:", tb_diem)
print("Trung bình cắt đầu-cuối:", tb_con_lai)
```

```python
#Bai 3.6
from collections import defaultdict

ds = [
    ("An", 9.0), ("Binh", 7.5), ("Chi", 8.6),
    ("Dung", 5.0), ("Em", 6.0), ("Phuc", 3.5),
    ("Giang", 7.0)
]

nhom = defaultdict(list)

for ten, diem in ds:
    if diem >= 8.5:
        chu = "A"
    elif diem >= 7:
        chu = "B"
    elif diem >= 5.5:
        chu = "C"
    elif diem >= 4:
        chu = "D"
    else:
        chu = "F"

    nhom[chu].append(ten)

for chu in sorted(nhom):
    print(chu, nhom[chu])
```

```python
#Bai 3.8
ch1 = {"ao": 120, "quan": 80, "non": 30}
ch2 = {"ao": 90, "giay": 150, "non": 20}

tong = ch1.copy()

for sp, tien in ch2.items():
    tong[sp] = tong.get(sp, 0) + tien

for sp, tien in sorted(tong.items(), key=lambda x: -x[1]):
    print(sp, tien)

print("Sp bán ở 2 cửa hàng:", ch1.keys() & ch2.keys())
print("Sp chỉ bán ở một cửa hàng:", ch1.keys() ^ ch2.keys())
```

```python
#Bai 3.10
n = 1000

nt = set(range(2, n + 1))

for p in range(2, int(n ** 0.5) + 1):
    if p in nt:
        for i in range(p * p, n + 1, p):
            nt.discard(i)

print("Số lượng:", len(nt))
print("Tổng:", sum(nt))

cap = []

for p in sorted(nt):
    if p + 2 in nt:
        cap.append((p, p + 2))

print("Số cặp sinh đôi:", len(cap))
print("Các cặp:", cap)
```

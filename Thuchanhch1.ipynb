```python
#Bài 1.1
import math

a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))
c = float(input("Nhập cạnh c: "))

# Điều kiện là một tam giác
if a + b > c and a + c > b and b + c > a:
# Chu vi tam giác
    p = (a + b + c) / 2
# Hệ thức Heron
    S = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print("Chu vi tam giác:", 2 * p)
    print("Diện tích tam giác:", S)
else:
    print("Ba cạnh không tạo thành tam giác!")
```

```python
#Bài 1.2
# Nhập số tiền gửi ban đầu
tien_gui = float(input("Nhập số tiền gửi ban đầu: "))

# Lãi suất năm 6%, quy đổi sang tháng
lai_suat = 0.06 / 12

# Tính số dư sau 1, 2 và 3 tháng
for thang in range(1, 4):
    tien_gui = tien_gui * (1 + lai_suat)
    print(f"Số dư sau {thang} tháng: {tien_gui:,.0f} đồng")
```

```python
#Bài 1.3

# Nhập tổng số phút
tong_phut = int(input("Nhập tổng số phút: "))

# Quy đổi sang ngày, giờ, phút
ngay = tong_phut // 1440
gio = (tong_phut % 1440) // 60
phut = tong_phut % 60

print(f"{ngay} ngày {gio} giờ {phut} phút")
```

```python
#Bài 1.4
# Số tiền cần rút
so_tien = 1_880_000

# Các mệnh giá tiền
menh_gia = [500000, 200000, 100000, 50000, 20000, 10000]
tong_da_tra = 0

for tien in menh_gia:
    so_to = so_tien // tien
    so_tien = so_tien % tien
    tong_da_tra += so_to * tien

    print(f"Tờ {tien:,.0f} đồng: {so_to} tờ")

# Số tiền còn lại
print(f"Số tiền còn lại: {so_tien:,.0f} đồng")

# Tổng tiền đã trả
print(f"Tổng tiền đã trả: {tong_da_tra:,.0f} đồng")
```
